## Vanishing Gradient Problem

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
x, y = make_moons(n_samples=250, noise=0.1,random_state=149)

x.shape
```

```python
plt.scatter(x[:,0], x[:,1], c=y)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=149)
```

```python
model = Sequential()

model.add(Dense(10, activation='sigmoid', input_dim=2))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
old_w = model.get_weights()[0]
old_w
```

```python
model.compile(loss='binary_crossentropy', optimizer='Adam', metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=25)
```

```python
new_w = model.get_weights()[0]
new_w
```

```python
lr = model.optimizer.get_config()['learning_rate']
lr
```

```python
gradient = (old_w - new_w)/lr
gradient
```

```python
grad_percentage_change = abs((old_w - new_w)) * 100 /old_w
grad_percentage_change
```
