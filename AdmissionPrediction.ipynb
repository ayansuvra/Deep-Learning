## Graduate Admission Prediction

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score

import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
df = pd.read_csv('./datasets/Admission_Predict_Ver1.1.csv')
df.head()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['Serial No.'], inplace=True)
```

```python
x, y = df.iloc[:,0:7], df.iloc[:,-1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=149, test_size=0.2)

scaler = MinMaxScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

### ANN

```python
model = Sequential()
model.add(Dense(7, activation='relu', input_dim=7))
model.add(Dense(3, activation='relu'))
model.add(Dense(1, activation='linear'))

model.summary()
```

```python
x_train.shape
```

```python
model.compile(optimizer='Adam', loss='mean_squared_error')
history = model.fit(x_train, y_train, validation_split=0.2, epochs=50, verbose=0)
```

```python
y_pred = model.predict(x_test)
```

```python
r2_score(y_test, y_pred)
```
